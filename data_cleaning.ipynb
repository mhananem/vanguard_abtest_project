# Clean Data

```python
df = df.rename(columns={
    'gendr': 'gender',
    'clnt_age':'client_age',
    'clnt_tenure_yr': 'client_tenure_yr',
    'clnt_tenure_mnth': 'client_tenure_month',
    'bal':'balance'
})
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
df.duplicated()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['client_tenure_yr','client_tenure_month','client_age','gender','num_accts','balance','calls_6_mnth','logons_6_mnth'] ,inplace=True)
```

```python
df.isnull().sum()
```

```python
df['client_age'] = df['client_age'].round().astype(int)
```

```python
df['num_accts'] = df['num_accts'].astype(int)
```

```python
df['calls_6_mnth'] = df['calls_6_mnth'].astype(int)
df['logons_6_mnth'] = df['logons_6_mnth'].astype(int)
```

```python
df.dtypes
```

```python
df['gender'] = df['gender'].replace({
    'U': 'Unknown',
    'M': 'Male',
    'F': 'Female'
})
```

```python
df.head()
```

```python
df.to_csv("cleaned_demo.csv",index=False)
```
