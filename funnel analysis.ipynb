##Step 1-Build Funnel Format

```python
funnel = df.groupby(['variation', 'process_step'])['visit_id'].nunique()
funnel = funnel.reset_index().pivot(index='process_step', columns='variation', values='visit_id')
```

```python
print(funnel)
```

##Order Standardized

```python
step_order = ['start', 'step_1', 'step_2', 'step_3', 'confirm']
funnel = funnel.loc[step_order]
```

##Step 2- Drop-off Rate(Where users leave)

```python
funnel['drop_off_control'] = funnel['Control'].pct_change().abs()*100
funnel['drop_off_test'] = funnel['Test'].pct_change().abs()*100
```

```python
funnel['drop_users_control'] = funnel['Control'].diff().abs()
funnel['drop_users_test'] = funnel['Test'].diff().abs()
```

```python
funnel.head()
```

Key Observation-

(Start-Step1)
Control drop-off: 23.82%
Test drop-off: 14.69%
                  The Test variation reduces early abandonment by ~9 percentage points, indicating a stronger first impression and smoother onboarding experiencence.

(Step_1 → Step_2)
Control: 14.48%
Test: 13.37%
                 Redesign is slightly improved.

(Step_2 → Step_3)
Control: 9.13%
Test: 9.46%
                 Drop-off remains almost unchanged
                 
(Step_3 → Confirm)
Control: 12.32%
Test: 2.05%
                Drop-off reduced by over 10 percentage points. Indicates significantly lower friction at the conversion stage

Final Insight
                Overall, the Test variation demonstrates a clear improvement in user retention across the funnel.The major reduction in drop-off at the final step suggests that the new design has user-friendly experience.


##Step 3: Conversion Rate (How efficiently users move)

1. Step-to-step

```python
funnel['conv_control'] = funnel['Control'] / funnel['Control'].shift(1)
funnel['conv_test'] = funnel['Test'] / funnel['Test'].shift(1)
```

```python
funnel['conv_control'] = (funnel['conv_control'] * 100).round(2)
funnel['conv_test'] = (funnel['conv_test'] * 100).round(2)
```

```python
funnel.head()
```

2.Overall conversion

```python
funnel['overall_conv_control'] = funnel['Control'].div(funnel.loc['start','Control']).mul(100).round(2)
funnel['overall_conv_test'] = funnel['Test'].div(funnel.loc['start','Test']).mul(100).round(2)
```

```python
funnel.head()
```

```python
funnel['conv_diff'] = funnel['overall_conv_test'] - funnel['overall_conv_control']
funnel['drop_diff'] = funnel['drop_off_control'] - funnel['drop_off_test']
```

```python
import pandas as pd

df1 = pd.read_csv('funnel_data.csv')

```

```python
# Reshape to long format
long_df = pd.DataFrame({
    'process_step': list(df1['process_step']) * 2,
    'variation':    ['Control'] * len(df1) + ['Test'] * len(df1),
    'users':        list(df1['Control']) + list(df1['Test']),
    'drop_off_pct': list(df1['drop_off_control']) + list(df1['drop_off_test']),
    'drop_users':   list(df1['drop_users_control']) + list(df1['drop_users_test']),
    'step_conv':    list(df1['conv_control']) + list(df1['conv_test']),
    'overall_conv': list(df1['overall_conv_control']) + list(df1['overall_conv_test']),
    'conv_diff':    list(df1['conv_diff']) * 2,
    'drop_diff':    list(df1['drop_diff']) * 2,
})

# Set step order
step_order = ['start', 'step_1', 'step_2', 'step_3', 'confirm']
long_df['step_order'] = long_df['process_step'].map({s: i for i, s in enumerate(step_order)})
long_df = long_df.sort_values('step_order')

long_df.to_csv('funnel_long.csv', index=False)
print(long_df)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# ── Your Data ────────────────────────────────────────────────────────────────
steps = ['Start', 'Step 1', 'Step 2', 'Step 3', 'Confirm']
control = [30910, 23548, 20138, 18300, 16046]
test    = [33157, 28285, 24503, 22186, 21731]
overall_conv_control = [100.00, 76.18, 65.15, 59.20, 51.91]
overall_conv_test    = [100.00, 85.31, 73.90, 66.91, 65.54]
drop_control = [float('nan'), 23.82, 14.48, 9.13, 12.32]
drop_test    = [float('nan'), 14.69, 13.37, 9.46,  2.05]

# ── Config ───────────────────────────────────────────────────────────────────
DARK_BLUE  = '#1B4F8A'
TEAL       = '#1A8FBB'
BG         = '#F4F9FE'
DROP_RED   = '#C0392B'
DROP_GREEN = '#1D7A4F'
max_users  = max(control[0], test[0])
max_width  = 4.5
n          = len(steps)
row_h      = 1.1
gap        = 0.14

fig, ax = plt.subplots(figsize=(14, 9))
fig.patch.set_facecolor(BG)
ax.set_facecolor(BG)

for i in range(n):
    top_w_ctrl = (control[i]   / max_users) * max_width
    bot_w_ctrl = (control[i+1] / max_users) * max_width if i < n-1 else top_w_ctrl * 0.85
    top_w_test = (test[i]      / max_users) * max_width
    bot_w_test = (test[i+1]    / max_users) * max_width if i < n-1 else top_w_test * 0.85

    y_top = (n - i) * row_h
    y_bot = (n - i - 1) * row_h + 0.1
    y_mid = (y_top + y_bot) / 2

    # ── Control trapezoid ────────────────────────────────────────────────
    ctrl_pts = np.array([
        [-gap - top_w_ctrl, y_top],
        [-gap,              y_top],
        [-gap,              y_bot],
        [-gap - bot_w_ctrl, y_bot],
    ])
    ax.add_patch(patches.Polygon(ctrl_pts, closed=True,
                                  facecolor=DARK_BLUE, edgecolor='white',
                                  linewidth=1.2, alpha=0.92, zorder=2))

    # ── Test trapezoid ───────────────────────────────────────────────────
    test_pts = np.array([
        [gap,              y_top],
        [gap + top_w_test, y_top],
        [gap + bot_w_test, y_bot],
        [gap,              y_bot],
    ])
    ax.add_patch(patches.Polygon(test_pts, closed=True,
                                  facecolor=TEAL, edgecolor='white',
                                  linewidth=1.2, alpha=0.92, zorder=2))

    # ── Labels inside bars ───────────────────────────────────────────────
    ax.text(-gap - top_w_ctrl * 0.5, y_mid,
            f"{control[i]:,}\n({overall_conv_control[i]:.1f}%)",
            ha='center', va='center', fontsize=9.5, fontweight='bold',
            color='white', zorder=3)

    ax.text(gap + top_w_test * 0.5, y_mid,
            f"{test[i]:,}\n({overall_conv_test[i]:.1f}%)",
            ha='center', va='center', fontsize=9.5, fontweight='bold',
            color='white', zorder=3)

    # ── Step label in centre ─────────────────────────────────────────────
    ax.text(0, y_mid, steps[i],
            ha='center', va='center', fontsize=9, fontweight='bold',
            color='#0D2B55', zorder=4,
            bbox=dict(boxstyle='round,pad=0.28', facecolor=BG,
                      edgecolor='#B8D3ED', linewidth=1))

    # ── Drop-off annotations between bands ───────────────────────────────
    if i < n - 1:
        y_gap = y_bot - 0.04   # sit in the gap between this and next band
        dc = drop_control[i+1]
        dt = drop_test[i+1]

        # Control drop-off (left side)
        ctrl_drop_x = -gap - top_w_ctrl * 0.5
        ax.annotate('', xy=(ctrl_drop_x, y_gap - 0.04),
                    xytext=(ctrl_drop_x, y_gap + 0.04),
                    arrowprops=dict(arrowstyle='->', color=DROP_RED,
                                   lw=1.4), zorder=5)
        ax.text(ctrl_drop_x, y_gap - 0.07,
                f"▼ {dc:.1f}% dropped",
                ha='center', va='top', fontsize=8, color=DROP_RED,
                fontweight='bold', zorder=5)

        # Test drop-off (right side)
        test_drop_x = gap + top_w_test * 0.5
        ax.annotate('', xy=(test_drop_x, y_gap - 0.04),
                    xytext=(test_drop_x, y_gap + 0.04),
                    arrowprops=dict(arrowstyle='->', color=DROP_RED,
                                   lw=1.4), zorder=5)
        ax.text(test_drop_x, y_gap - 0.07,
                f"▼ {dt:.1f}% dropped",
                ha='center', va='top', fontsize=8, color=DROP_RED,
                fontweight='bold', zorder=5)

        # Diff badge in centre — green if test dropped less, red if more
        diff = dc - dt   # positive = test retained more
        diff_color = DROP_GREEN if diff > 0 else DROP_RED
        diff_label = f"Test saves\n+{diff:.1f}%" if diff > 0 else f"Test loses\n{diff:.1f}% more"
        ax.text(0, y_gap - 0.05,
                diff_label,
                ha='center', va='top', fontsize=7.5,
                color=diff_color, fontweight='bold', zorder=5,
                bbox=dict(boxstyle='round,pad=0.25',
                          facecolor='white', edgecolor=diff_color,
                          linewidth=1, alpha=0.9))

# ── Column headers ────────────────────────────────────────────────────────────
ax.text(-max_width * 0.5, n * row_h + 0.08,
        '◀  CONTROL', ha='center', va='bottom',
        fontsize=11, fontweight='bold', color=DARK_BLUE)
ax.text(max_width * 0.5, n * row_h + 0.08,
        'TEST  ▶', ha='center', va='bottom',
        fontsize=11, fontweight='bold', color=TEAL)

# ── Title ─────────────────────────────────────────────────────────────────────
ax.text(0, n * row_h + 0.62,
        'Funnel Analysis — Test vs Control',
        ha='center', va='center', fontsize=16,
        fontweight='bold', color='#0D2B55')
ax.text(0, n * row_h + 0.30,
        'Overall conversion:  Control 51.91%   ·   Test 65.54%   ·   Lift  +13.63%',
        ha='center', va='center', fontsize=10,
        color='#2575B8', style='italic')

# ── Legend ────────────────────────────────────────────────────────────────────
legend_elements = [
    patches.Patch(facecolor=DARK_BLUE, label='Control', alpha=0.92),
    patches.Patch(facecolor=TEAL,      label='Test',    alpha=0.92),
    patches.Patch(facecolor='white', edgecolor=DROP_GREEN,
                  label='Test saves (lower drop-off)'),
    patches.Patch(facecolor='white', edgecolor=DROP_RED,
                  label='Test loses more'),
]
ax.legend(handles=legend_elements, loc='lower center',
          bbox_to_anchor=(0.5, -0.03), ncol=4,
          fontsize=9.5, frameon=True, framealpha=0.9,
          edgecolor='#B8D3ED', facecolor=BG)

# ── Axes ──────────────────────────────────────────────────────────────────────
ax.set_xlim(-max_width - 1.0, max_width + 1.0)
ax.set_ylim(-0.5, n * row_h + 0.9)
ax.axis('off')

plt.tight_layout()
plt.savefig('funnel_chart.png', dpi=150, bbox_inches='tight', facecolor=BG)
plt.show()
print("Saved → funnel_chart.png")
```
