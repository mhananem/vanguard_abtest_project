###    1. Completion rate
**“Is the completion rate of the new process the same as the old process?”**
- H0  : completion rate (Test) =< completion  rate (control)
- H1  : completion rate (Test) > completion rate (control)

```python
visit_completed= pd.read_csv("clean_data/visit_completed.csv")
```

```python
visit_completed['completed'].dtype
```

```python
test_group = visit_completed[visit_completed['variation'] == 'Test']['completed']
control_group = visit_completed[visit_completed['variation'] == 'Control']['completed']
```

```python
alpha = 0.05
```

```python
_, p_value = stats.ttest_ind(test_group, control_group, equal_var=False)
print("p-value:", p_value)
```

**Conclusion**
- p_value < alpha (0.05) ==> We reject H0 and accept H1
- Completion rate (test) is Higher than completion rate (control)

###    1.a Completion rate - cost effectiveness
***Was the completion rate cost effective  (meaning higher than the 5% threshhold)?***

```python
#Absolut lift
lift = test_group.mean() - control_group.mean()
print(f"Observed lift: {lift:.2%}")
print(test_group.mean())
print(control_group.mean())

if lift >= 0.05:
    print("Meets the 5% threshold ✅")
else:
    print("Does not meet the 5% threshold ❌")
```

###    3. Average age of clients
**“Is the average age of clients engaging with the new process the same as those using the old process?”**
- H0​ : μtest ​= μcontrol​
- H1 : μtest​ ≠ μcontrol​

```python
df=pd.read_csv("clean_data/demo_exp_merged.csv")
```

```python
df.head()
```

```python

df = df.dropna(subset=['client_age'])
```

```python
test_age = df[df['Variation'] == 'Test']['client_age']
control_age = df[df['Variation'] == 'Control']['client_age']
```

```python
_, p_value = stats.ttest_ind(test_age, control_age
                          , equal_var=False)
print(p_value)
```

the p value is lower than alpha. that means we reject null hypothesis that mean age of test and control are not equal.

###    4. Average clients tenure
**Is the average tenure of clients engaging with the new process the same as those using the old process?**
- H0​ : tenure for test and control are same
- H1 : tenure for test and control are different

```python
df = df.dropna(subset=['client_tenure_month'])
```

```python
test_tenure = df[df['Variation'] == 'Test']['client_tenure_month']
control_tenure = df[df['Variation'] == 'Control']['client_tenure_month']
```

```python
_, p_value = stats.ttest_ind(test_tenure, control_tenure
                          , equal_var=False)
print(p_value)
```

p value is greater than alpha. we cannot reject H0. it means that tenure for test and control is same.

###    5. Clients Gender
**Is the average gender of clients engaging with the new process the same as those using the old process?**
- H0​ : gender distribution is independent of variations (test/control)
- H1 : gender distribution differs accross variations

```python
import pandas as pd

contingency = pd.crosstab(df['Variation'], df['gender'])
print(contingency)
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(contingency)

print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value <= alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

There is no statistically significant association between gender and variation (Test vs Control).
