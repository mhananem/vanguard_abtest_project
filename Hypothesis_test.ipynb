5.## Gender for test and control.

H0: Gender distribution is independent of variation (Test/Control)

H1: Gender distribution differs across variations

```python
import pandas as pd

contingency = pd.crosstab(df1['Variation'], df1['gender'])
print(contingency)
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(contingency)

print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value <= alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

There is no statistically significant association between gender and variation (Test vs Control).
